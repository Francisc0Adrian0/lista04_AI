```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
```

```python
dados = pd.read_csv("acessos_sistema.csv")

dados.head()
```

```python
dados.info()
```

```python
dados.describe()
```

```python
dados.hist(figsize=(12,8))
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    dados.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Matriz de Correlação")

plt.show()
```

```python
scaler = StandardScaler()

dados_normalizados = scaler.fit_transform(dados)
```

```python
modelo = IsolationForest(
    contamination=0.10,
    random_state=42
)

modelo.fit(dados_normalizados)
```

```python
dados["anomalia"] = modelo.predict(dados_normalizados)
```

```python
dados["anomalia"].value_counts()
```

```python
dados["anomalia"].value_counts()
```

```python
anomalias = dados[dados["anomalia"] == -1]

anomalias
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=dados,
    x="duracao_sessao_min",
    y="quantidade_paginas_acessadas",
    hue="anomalia",
    palette="Set1"
)

plt.title("Detecção de Anomalias")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(data=dados.drop("anomalia", axis=1))

plt.xticks(rotation=45)

plt.show()
```

O algoritmo Isolation Forest foi utilizado para identificar acessos potencialmente anômalos em logs de sistema. Os registros classificados como anomalias apresentaram características incomuns em relação ao comportamento predominante dos usuários, como horários atípicos de acesso, sessões muito longas, utilização de IP diferente do habitual e grande quantidade de páginas acessadas. Os resultados indicam possíveis eventos que merecem investigação adicional, embora nem toda anomalia represente necessariamente um incidente de segurança.
